```python
## Import packages

import xarray as xr

## Load in data

ds_cGAN = xr.open_dataset('../datasets/east_africa-cgan_forecast-20240422.nc')
```

## What is in the data?

```python
## What does the data look like?

ds_cGAN
```

```python
## Check which members are available to select

ds_cGAN.member.values
```

```python
## Select a given member or set of members using sel (i.e., by index label)

ds_cGAN.sel({'member':[24]})
```

```python
## Select the first lon and lat point using isel (i.e. by index integer)

ds_cGAN.isel({'longitude':[0], 'latitude':[0]})
```

#### Note this would be the same as doing ds_cGAN.sel({'longitude':[19.15], 'latitude':[-13.65]})!
